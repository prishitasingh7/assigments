## 12. Stacking & Concatenation

_`np.vstack()` → stack rows | `np.hstack()` → stack columns | `np.concatenate(axis=)` → general version_


**Q12.1** Create two 2x2 arrays. Combine them using vstack, hstack, and concatenate on both axis=0 and axis=1.

```python
# Write your code here

import numpy as np

# Create two 2x2 arrays
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# vstack
v_result = np.vstack((A, B))

# hstack
h_result = np.hstack((A, B))

# concatenate along axis=0
concat_axis0 = np.concatenate((A, B), axis=0)

# concatenate along axis=1
concat_axis1 = np.concatenate((A, B), axis=1)

print("A:")
print(A)

print("B:")
print(B)

print("vstack:")
print(v_result)

print("hstack:")
print(h_result)

print("concatenate axis=0:")
print(concat_axis0)

print("concatenate axis=1:")
print(concat_axis1)
```

## 14. Matrix Operations & Transpose

_`*` → element-wise multiply | `@` / `np.dot()` → matrix multiplication | `.T` → transpose (rows ↔ cols)_


**Q14.1** Create two 2x2 matrices. Compute their element-wise product and their matrix product (using both @ and np.dot).

```python
# Write your code here
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
element_wise = A * B
matrix_product_at = A @ B
matrix_product_dot = np.dot(A, B)
print("Matrix A:")
print(A)
print("\nMatrix B:")
print(B)
print("\nElement-wise product:")
print(element_wise)
print("\nMatrix product using @:")
print(matrix_product_at)
print("\nMatrix product using np.dot():")
print(matrix_product_dot)
```

**Q14.2** Print the transpose of the first matrix.

```python
# Write your code here
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

transpose = [list(row) for row in zip(*matrix)]

print("Transpose of first matrix:")
for row in transpose:
    print(row)
```
