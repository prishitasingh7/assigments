```python
cat_cols = ['Sex', 'Embarked']
num_cols = ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(
    [('cat', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), cat_cols)],
    remainder='passthrough'
)
```

```python
x_train_final = ct.fit_transform(x_train[cat_cols + num_cols])
x_test_final = ct.transform(x_test[cat_cols + num_cols])
```

```python
model = KNeighborsClassifier()

model.fit(x_train_final, y_train)

y_pre = model.predict(x_test_final)

print(accuracy_score(y_test, y_pre))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_final_scaled = scaler.fit_transform(x_train_final)
x_test_final_scaled = scaler.transform(x_test_final)
```

```python
scaled_model = KNeighborsClassifier()

scaled_model.fit(x_train_final_scaled, y_train)

y_pre_scaled = scaled_model.predict(x_test_final_scaled)

print(accuracy_score(y_test, y_pre_scaled))
```

```python
scaled_model = KNeighborsClassifier()

scaled_model.fit(x_train_final_scaled, y_train)

y_pre_scaled = scaled_model.predict(x_test_final_scaled)

print(accuracy_score(y_test, y_pre_scaled))
```

```python
for k in [3,5,7,11,13,15,21,25,29]:
    m = KNeighborsClassifier(n_neighbors=k)
    m.fit(x_train_final_scaled, y_train)
    m_pred = m.predict(x_test_final_scaled)
    print(k, round(accuracy_score(y_test, m_pred), 2))
```
